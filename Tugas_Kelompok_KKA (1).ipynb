```python
import pandas as pd

# Memuat dataset global ecommerce sales
df = pd.read_csv('global_ecommerce_sales.csv')

print("--- 5 Baris Pertama ---")
print(df.head()) # Lihat 5 baris pertama

print("\n--- Informasi Data & Missing Value ---")
print(df.info()) # Tipe data & jumlah data kosong

print("\n--- Ringkasan Statistik ---")
print(df.describe()) # Ringkasan statistik

# Rata-rata total pembelian per kategori produk
rata_kategori = df.groupby('Category')['Total_Amount'].mean()
print("\n--- Rata-rata Total Pembelian per Kategori ---")
print(rata_kategori.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

total_kategori = df.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=total_kategori.values, y=total_kategori.index, palette='Blues_r')
plt.title('Total Penjualan per Kategori Produk')
plt.xlabel('Total Penjualan ($)')
plt.ylabel('Kategori Produk')
plt.tight_layout()
plt.show()
```

Jawaban Tugas Analisis 2:

Penyumbang Penjualan: Kategori Home Decor menyumbang penjualan tertinggi (sekitar $232,275), sedangkan Books menyumbang terendah (sekitar $158,579)

Rekomendasi Bisnis: Fokuskan strategi promosi dan inventaris pada kategori unggulan (Home Decor, Grocery, dan Electronics), serta evaluasi penetapan harga atau campaign promosi untuk meningkatkan performa kategori Books.

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['bulan'] = df['Order_Date'].dt.to_period('M')
tren_bulanan = df.groupby('bulan')['Total_Amount'].sum()

plt.figure(figsize=(8,5))
plt.plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Jawaban Tugas Analisis 4:

Pola Hubungan: Terlihat hubungan positif; semakin banyak kuantitas barang yang dibeli, nilai Total_Amount cenderung semakin tinggi.

Outlier/Pencilan: Beberapa titik data pada kuantitas tinggi (misal Quantity = 9 atau 10) memiliki Total_Amount hingga mendekati $5,000, terutama pada kategori Electronics dan Fashion.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca dataset
df = pd.read_csv('global_ecommerce_sales.csv')

# Mengelompokkan data harga satuan menjadi 5 kategori
kategori_harga = pd.cut(df['Unit_Price'], bins=5).value_counts().sort_index()
labels = [f"${interval.left:.2f} - ${interval.right:.2f}" for interval in kategori_harga.index]

# Membuat diagram lingkaran
plt.figure(figsize=(7, 5))
plt.pie(
    kategori_harga,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.light_palette('seagreen', n_colors=5)
)
plt.title('Distribusi Harga Satuan (Unit Price)')
plt.axis('equal')
plt.tight_layout()
plt.savefig('pie_final.png', dpi=150)
plt.show()

# Boxplot jumlah kuantitas pembelian per kategori (deteksi outlier)
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Category', y='Quantity')
plt.title('Sebaran Kuantitas Pembelian per Kategori Produk')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

Jawaban Tugas Analisis 5:

Distribusi Harga Satuan: Distribusi Unit_Price relatif merata (uniform) melingkupi rentang $10 hingga $500.

Boxplot & Outlier: Kuantitas pembelian tersebar cukup merata antara 1 hingga 10 unit pada tiap kategori produk tanpa outlier ekstrem pada variabel Quantity.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Membaca & menyiapkan data
df = pd.read_csv('global_ecommerce_sales.csv.xls')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

total_kategori = df.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False)
tren_bulanan = df.groupby(df['Order_Date'].dt.to_period('M'))['Total_Amount'].sum()

# Mengelompokkan harga satuan menjadi 5 kategori untuk Pie Chart
kategori_harga = pd.cut(df['Unit_Price'], bins=5).value_counts().sort_index()
labels_harga = [f"${interval.left:.2f} - ${interval.right:.2f}" for interval in kategori_harga.index]

# 2. Membuat Dashboard 2x2
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# Grafik 1: Barplot Total Penjualan per Kategori
sns.barplot(x=total_kategori.values, y=total_kategori.index, ax=axes[0,0], palette='Blues_r')
axes[0,0].set_title('Total Penjualan per Kategori')
axes[0,0].set_xlabel('Total Penjualan ($)')

# Grafik 2: Lineplot Tren Bulanan
axes[0,1].plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
axes[0,1].set_title('Tren Penjualan Bulanan')
axes[0,1].tick_params(axis='x', rotation=45)

# Grafik 3: Diagram Lingkaran Distribusi Unit Price (menggantikan histplot)
axes[1,0].pie(
    kategori_harga,
    labels=labels_harga,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.light_palette('seagreen', n_colors=5)
)
axes[1,0].set_title('Distribusi Harga Satuan (Unit Price)')

# Grafik 4: Boxplot Sebaran Quantity per Kategori
sns.boxplot(data=df, x='Category', y='Quantity', ax=axes[1,1])
axes[1,1].set_title('Sebaran Kuantitas per Kategori')
axes[1,1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.savefig('dashboard_final.png', dpi=150)
plt.show()
```

Jawaban Tugas Analisis 6:

Keuntungan Dashboard: Memudahkan pengambil keputusan melihat beberapa sudut pandang performa bisnis secara ringkas dan bersamaan tanpa berpindah halaman.

Keputusan Bisnis: Mengalokasikan stok lebih banyak pada bulan pertengahan tahun (Juli) dan memfokuskan promosi pada kategori Home Decor dan Grocery
